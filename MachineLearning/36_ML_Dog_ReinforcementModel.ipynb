```python
import random
```

```python
# Actions that the dog can take
actions=['sit','stand']
```

```python
# allot memory for rewards for each action
reward_table={
    "sit":0,
    "stand":0
}
```

**Training Loop**

```python
# I instruct the dog to SIT
for episode in range(1,21):
  # Dog chooses action randomly
  action=random.choice(actions)

  #Give reward for correct action
  if action=="sit":
    reward=1  # biscuit
  else:
    reward=0

  reward_table[action]+=reward
  print(f"Episode {episode}: Action : {action}, Reward: {reward}")

print("Learning result (Reward table) :",reward_table)
```

```python
best_action=max(reward_table,key=reward_table.get)
print("Dog has learned to: ",best_action)
```
