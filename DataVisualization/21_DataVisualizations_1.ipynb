```python
import pandas
#import matplotlib.pyplot as plt
from matplotlib import pyplot
```

**LINE Chart-Growth & Decline**

```python
data={
    "month":["Dec25","Jan26","Feb26","Mar26","Apr26","May26"],
    "sales":[20000,25000,17500,30000,21000,19700]
}
df=pandas.DataFrame(data)
print(df)
```

```python
pyplot.plot(df["month"],df["sales"],marker="o")
pyplot.xlabel("Month")
pyplot.ylabel("Sale")
pyplot.title("Sales in last 6 months")
pyplot.show()
```

**BAR chart - compare vaules**

```python
product=['laptop','mobile','tablet','smartwatch']
sale=[50,110,27,65]

pyplot.bar(product,sale)
pyplot.xlabel("Product")
pyplot.xlabel("Sale")
pyplot.title("Items sold in April 2026")
pyplot.show()
```

**PIE Chart - percentage distribution**

```python
product=['laptop','mobile','tablet','smartwatch']
sale=[50,110,27,65]

pyplot.pie(sale,labels=product,autopct='%1.1f%%')
pyplot.title("Sales share productwise")
pyplot.show()
```

**SCATTER chart - identifies correlation**

```python
study_hours=[1,2,3,4,5]
marks=[40,50,60,70,90]

pyplot.scatter(study_hours,marks)
pyplot.title("Study hours vs marks")
pyplot.show()
```
