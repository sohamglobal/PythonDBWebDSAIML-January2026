```python
import numpy
```

**Filter**

```python
arr=numpy.random.randint(1,75,10)
print(arr)
```

```python
filtered=arr[arr>50]
print(filtered)
```

```python
print(arr[arr%2==1])
print(arr[arr%2==0])
```

**Sort**

```python
arr=numpy.random.randint(25,100,10)
print(arr)
```

```python
print(numpy.sort(arr))
```

```python
print(numpy.sort(arr)[::-1])
```

**Unique**

```python
arr=numpy.array([9,26,13,45,1,9,25,29,1,13,9,26,45])
print(arr)
```

```python
un=numpy.unique(arr)
print(un)
```

**Split**

```python
arr=numpy.array([9,26,13,45,1,9,25,29,1,13,9,26,45,13,1])
print(arr.shape)
print(numpy.split(arr,5))
```

```python
arr=numpy.array([9,26,13,45,1,9,25,29,1,13,9,26])
print(arr.shape)
print(numpy.split(arr,5))
```

```python
arr=numpy.array([9,26,13,45,1,9,25,29,1,13,9,26])
print(arr.shape)
print(numpy.array_split(arr,5))
```
